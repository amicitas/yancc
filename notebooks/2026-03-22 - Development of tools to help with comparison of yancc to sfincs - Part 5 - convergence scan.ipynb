# Convergence Scan Demonstration
This notebook demonstrates the automated convergence scan tool for the Yancc solver.
It iterates over numerical resolution parameters and tracks particle fluxes.

```python
import numpy as np
import logging
from pathlib import Path

import desc
import yancc
from yancc.species import GlobalMaxwellian
from yancc import yancctools
from yancc.yancctools import run_convergence_scan, plot_convergence_scan, save_results, load_results

logging.basicConfig(level=logging.INFO, force=True)
```

```python
# 1. Setup Basic Equilibrium and Profiles
eq = desc.examples.get("W7-X")

def density(r): return 1.5e20 * (1 - r**2)
def temp_e(r): return 3.0e3 * (1 - r**2)
def temp_i(r): return 1.5e3 * (1 - r**2)

global_species = [
    GlobalMaxwellian(yancc.species.Electron, temperature=temp_e, density=density),
    GlobalMaxwellian(yancc.species.Hydrogen, temperature=temp_i, density=density)
]
```

```python
# 2. Define Baseline Options
options = {
    "nx": 5,   # Speed resolution
    "na": 35,  # Pitch-angle resolution
    "nt": 13,  # Poloidal resolution
    "nz": 25,  # Toroidal resolution
    "convergence_scan_steps": 3,  # Points per parameter to scan
    "rtol": 1e-3,
    "output_path": "./runs"
}
```

```python
# 3. Execute Convergence Scan
results = run_convergence_scan(
    eq_type="desc",
    eq_data=eq,
    global_species=global_species,
    rho=0.5,
    erho=0.0,
    options=options
)

runid = results["runid"]
print(f"Convergence scan completed. RunID: {runid}")
```

```python
# 4. Save Results (Optional)
run_dir = Path(options["output_path"]) / runid
run_dir.mkdir(parents=True, exist_ok=True)
save_results(results, run_dir)
```

```python
# 5. Plot Results
fig = plot_convergence_scan(results)
```
